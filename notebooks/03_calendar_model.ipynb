```python
from pathlib import Path
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

project_dir = Path.cwd()
if not (project_dir / "src" / "preparation.py").exists():
    project_dir = project_dir.parent
assert (project_dir / "src" / "preparation.py").exists()
src_dir = str(project_dir / "src")
if src_dir not in sys.path:
    sys.path.insert(0, src_dir)
from preparation import (
    load_quarter_hour_data,
    prepare_quarter_hour_data,
    aggregate_hourly_energy,
)
```

## 1. Load and validate both years

Each source hour must contain four quarter-hour observations. Check continuity across the year boundary before using row shifts as elapsed-hour lags.

```python
hourly_frames = []
for year in [2023, 2024]:
    raw = load_quarter_hour_data(
        project_dir / "data" / "raw" / f"EnergieUebersichtCH-{year}.xlsx"
    )
    prepared = prepare_quarter_hour_data(raw)
    frame = aggregate_hourly_energy(prepared)
    assert len(frame) == (8784 if year == 2024 else 8760)
    assert frame["interval_count"].eq(4).all()
    assert frame["end_user_energy_mwh"].notna().all()
    assert np.isclose(
        raw["end_user_energy_kwh"].sum(), frame["end_user_energy_kwh"].sum(),
        rtol=1e-12, atol=0.01,
    )
    hourly_frames.append(frame)

hourly = pd.concat(hourly_frames)
assert hourly.index.is_unique
assert hourly.index.is_monotonic_increasing
assert hourly.index.to_series().diff().dropna().eq(pd.Timedelta(hours=1)).all()
print(f"Prepared {len(hourly):,} complete hours across 2023 and 2024.")
```

## 2. Freeze the training set before the first forecast

For each hourly observation, calculate maturity from its interval end. Training includes only values mature by the first issue time. This is stricter than checking that a timestamp is merely earlier than its target.

The evaluation window and noon issue times match notebook 02. Training remains fixed, even when more observations would mature later in the quarter.

```python
def assumed_maturity_utc(interval_starts):
    local_ends = (
        interval_starts + pd.Timedelta(hours=1)
    ).tz_convert("Europe/Zurich").tz_localize(None)
    maturity = (
        local_ends.to_period("M").to_timestamp()
        + pd.offsets.MonthBegin(1)
        + pd.DateOffset(months=6)
    )
    return maturity.tz_localize("Europe/Zurich").tz_convert("UTC")

start_local = pd.Timestamp("2024-10-01", tz="Europe/Zurich")
end_local = pd.Timestamp("2025-01-01", tz="Europe/Zurich")
evaluation = hourly.loc[
    (hourly["hour_start_local"] >= start_local)
    & (hourly["hour_start_local"] < end_local),
    ["end_user_energy_mwh", "hour_start_local"],
].rename(columns={"end_user_energy_mwh": "actual_mwh"}).copy()
local_dates = evaluation["hour_start_local"].dt.tz_localize(None).dt.normalize()
evaluation["issue_time_utc"] = (
    local_dates - pd.Timedelta(days=1) + pd.Timedelta(hours=12)
).dt.tz_localize("Europe/Zurich").dt.tz_convert("UTC")
assert len(evaluation) == 2209
first_issue = evaluation["issue_time_utc"].min()

maturity = assumed_maturity_utc(hourly.index)
train_mask = (maturity <= first_issue) & (hourly.index < first_issue)
training = hourly.loc[train_mask].copy()
assert len(training) > 0
assert (maturity[train_mask] <= first_issue).all()
assert training.index.intersection(evaluation.index).empty

print("First issue:", first_issue.tz_convert("Europe/Zurich"))
print("Training starts:", training["hour_start_local"].iloc[0])
print("Last training interval starts:", training["hour_start_local"].iloc[-1])
print(f"Training hours: {len(training):,}; evaluation hours: {len(evaluation):,}")
```

## 3. Calendar features and least-squares fit

Use 168 indicator columns for local weekday/hour combinations (Monday 00:00 through Sunday 23:00). No extra intercept is needed: exactly one indicator is active per row. Two additional columns represent a smooth annual sine/cosine cycle, using each year's actual length.

The annual correction is shared by all weekday/hour combinations. This intentionally simple model has no weather, holiday, recent-consumption, or trend inputs. It predicts the same energy for both occurrences of a repeated local hour. All features are known before issue time.

`np.linalg.lstsq` chooses coefficients to minimize squared training errors. It requires no additional modelling package.

```python
def calendar_features(local_times):
    local_times = pd.DatetimeIndex(local_times)
    hour_of_week = local_times.dayofweek * 24 + local_times.hour
    weekly_profile = np.eye(168)[hour_of_week]
    days_in_year = np.where(local_times.is_leap_year, 366, 365)
    year_fraction = (
        local_times.dayofyear - 1 + local_times.hour / 24
    ) / days_in_year
    angle = 2 * np.pi * np.asarray(year_fraction)
    return np.column_stack([
        weekly_profile, np.sin(angle), np.cos(angle),
    ])

X_train = calendar_features(training["hour_start_local"])
y_train = training["end_user_energy_mwh"].to_numpy()
coefficients, _, rank, _ = np.linalg.lstsq(X_train, y_train, rcond=None)
assert rank == X_train.shape[1]

X_eval = calendar_features(evaluation["hour_start_local"])
evaluation["calendar_model_mwh"] = X_eval @ coefficients
assert np.isfinite(evaluation["calendar_model_mwh"]).all()
print(f"Fitted {len(coefficients)} coefficients once on mature training data.")
```

## 4. Compare with the unchanged 52-week baseline

Both methods use the same target hours. Check that the 52-week source intervals also meet the maturity policy at their respective issue times. Metrics weight each evaluated hour equally; mean error is prediction minus actual.

```python
lag = pd.Timedelta(weeks=52)
evaluation["52_week_prediction_mwh"] = (
    hourly["end_user_energy_mwh"].shift(52 * 168).reindex(evaluation.index)
)
source_maturity = assumed_maturity_utc(evaluation.index - lag)
assert (source_maturity <= evaluation["issue_time_utc"]).all()
prediction_columns = ["52_week_prediction_mwh", "calendar_model_mwh"]
assert evaluation[["actual_mwh"] + prediction_columns].notna().all().all()

errors = evaluation[prediction_columns].sub(evaluation["actual_mwh"], axis=0)
scores = pd.DataFrame({
    "MAE (MWh)": errors.abs().mean(),
    "RMSE (MWh)": np.sqrt(errors.pow(2).mean()),
    "Mean error (MWh)": errors.mean(),
})
display(scores.round(2))

monthly_mae = (
    errors.abs()
    .assign(local_month=evaluation["hour_start_local"].dt.month)
    .groupby("local_month").mean()
)
display(monthly_mae.round(2))
```

## 7. Add one predefined holiday indicator

Test one extra feature: 1 on January 1 or December 25-26, otherwise 0, using the **Swiss local interval-start date**. These dates are an exploratory definition, not a complete Swiss regional holiday calendar. The rule was chosen after inspecting development errors; it is not an independent test.

Append the indicator to the original 170 calendar features and refit all 171 coefficients by ordinary least squares on exactly the same mature training rows. The new coefficient is one shared additive correction in MWh for every marked hour; it cannot learn a separate holiday daily shape. Keep both original prediction columns and the original coefficients unchanged. Freeze the new model at the same first issue time.

First show how many distinct marked training dates support the coefficient. Then compare all three methods across the complete quarter and each month, followed by marked and unmarked hours. January 1 is present in training but absent from this evaluation quarter, so its forecast performance is not tested here. No further feature selection is performed in this experiment.

```python
def holiday_indicator(local_times):
    local_times = pd.DatetimeIndex(local_times).tz_convert("Europe/Zurich")
    return np.asarray(
        ((local_times.month == 1) & (local_times.day == 1))
        | ((local_times.month == 12) & np.isin(local_times.day, [25, 26])),
        dtype=float,
    )

# UTC timestamps must be classified by their Swiss local date.
rule_probe = pd.to_datetime([
    "2023-12-24 22:00Z", "2023-12-24 23:00Z",
    "2023-12-26 22:00Z", "2023-12-26 23:00Z",
    "2023-12-31 23:00Z", "2024-01-01 23:00Z",
    "2024-07-31 22:00Z",
])
np.testing.assert_array_equal(holiday_indicator(rule_probe), [0, 1, 1, 0, 1, 0, 0])

holiday_train = holiday_indicator(training["hour_start_local"])
holiday_eval = holiday_indicator(evaluation["hour_start_local"])
training_holiday_counts = (
    training.loc[holiday_train.astype(bool)]
    .groupby(training.loc[holiday_train.astype(bool), "hour_start_local"].dt.date)
    .size().rename("Training hours")
)
display(training_holiday_counts.to_frame())
print(f"Marked training dates: {len(training_holiday_counts)}; "
      f"marked training hours: {int(holiday_train.sum())}; "
      f"marked evaluation hours: {int(holiday_eval.sum())}")

original_predictions = evaluation[prediction_columns].copy(deep=True)
original_coefficients = coefficients.copy()
X_train_holiday = np.column_stack([X_train, holiday_train])
X_eval_holiday = np.column_stack([X_eval, holiday_eval])
assert len(training) == 10199 and len(evaluation) == 2209
assert X_train_holiday.shape == (len(training), 171)
assert (maturity[train_mask] <= first_issue).all()
holiday_coefficients, _, holiday_rank, _ = np.linalg.lstsq(
    X_train_holiday, y_train, rcond=None,
)
assert holiday_rank == X_train_holiday.shape[1]
evaluation["calendar_holiday_model_mwh"] = X_eval_holiday @ holiday_coefficients
assert np.isfinite(evaluation["calendar_holiday_model_mwh"]).all()
pd.testing.assert_frame_equal(evaluation[prediction_columns], original_predictions)
np.testing.assert_array_equal(coefficients, original_coefficients)
print(f"Fitted {len(holiday_coefficients)} coefficients on unchanged training rows.")
print(f"Shared holiday coefficient: {holiday_coefficients[-1]:+.2f} MWh")
```

```python
holiday_prediction_columns = prediction_columns + ["calendar_holiday_model_mwh"]
holiday_errors = evaluation[holiday_prediction_columns].sub(evaluation["actual_mwh"], axis=0)

def error_metrics(error_frame):
    return pd.DataFrame({
        "MAE (MWh)": error_frame.abs().mean(),
        "RMSE (MWh)": np.sqrt(error_frame.pow(2).mean()),
        "Mean error (MWh)": error_frame.mean(),
    })

holiday_scores = error_metrics(holiday_errors)
print("Whole quarter (2,209 hours):")
display(holiday_scores.round(2))
holiday_monthly_scores = pd.concat({
    month: error_metrics(month_errors)
    for month, month_errors in holiday_errors.groupby(evaluation["hour_start_local"].dt.month)
}, names=["Swiss local month", "Model"])
print("Monthly comparison:")
display(holiday_monthly_scores.round(2))
holiday_subset_scores = pd.concat({
    label: error_metrics(holiday_errors.loc[mask])
    for label, mask in [
        ("Marked dates (48 hours)", holiday_eval.astype(bool)),
        ("Other dates (2,161 hours)", ~holiday_eval.astype(bool)),
    ]
}, names=["Subset", "Model"])
print("Diagnostic subsets (each hour weighted equally):")
display(holiday_subset_scores.round(2))
mae_gain = 100 * (1 - holiday_scores.loc["calendar_holiday_model_mwh", "MAE (MWh)"]
                     / holiday_scores.loc["calendar_model_mwh", "MAE (MWh)"])
print(f"Whole-quarter MAE improvement over original calendar model: {mae_gain:.2f}%")

# Refit changes the weekly and annual coefficients too, including predictions
# on unmarked dates; the extra column is not a post-hoc forecast subtraction.
print("Refitting also changes predictions on unmarked dates.")
```

```python
holiday_window = evaluation.loc[
    (evaluation["hour_start_local"] >= pd.Timestamp("2024-12-23", tz="Europe/Zurich"))
    & (evaluation["hour_start_local"] < pd.Timestamp("2024-12-28", tz="Europe/Zurich"))
]
fig, ax = plt.subplots(figsize=(12, 4))
for column, label in [
    ("actual_mwh", "Actual"),
    ("52_week_prediction_mwh", "52-week baseline"),
    ("calendar_model_mwh", "Original calendar model"),
    ("calendar_holiday_model_mwh", "Calendar + holiday indicator"),
]:
    ax.plot(holiday_window["hour_start_local"], holiday_window[column], label=label)
ax.axvspan(pd.Timestamp("2024-12-25", tz="Europe/Zurich"),
           pd.Timestamp("2024-12-27", tz="Europe/Zurich"),
           color="grey", alpha=0.12, label="Marked dates")
ax.xaxis.set_major_formatter(mdates.DateFormatter("%m-%d %H:%M", tz=ZoneInfo("Europe/Zurich")))
ax.set(xlabel="Swiss local hour start", ylabel="Energy per hour (MWh)",
       title="Holiday experiment: December 23-27, 2024 (development diagnostic)")
ax.legend()
ax.grid(alpha=0.3)
fig.autofmt_xdate()
plt.tight_layout()
plt.show()
```

## 8. Training-only holiday coefficient sensitivity

Leave out each of the four marked Swiss local dates in turn, removing **all 24 hours** of that date from the original mature training set. Refit all 171 coefficients on the remaining rows, retaining the same feature definition and maturity cutoff. All unmarked training dates remain in every fit.

Compare each holiday coefficient with the full-training estimate. This checks dependence on individual dates; it is not a confidence interval or a chronological forecasting backtest. Some retained training dates occur after the omitted date. No evaluation-period predictions or errors enter this diagnostic, and no fitted model is replaced or selected using its results.


```python
sensitivity_training_dates = training["hour_start_local"].dt.date.to_numpy()
sensitivity_marked_dates = sorted(set(sensitivity_training_dates[holiday_train.astype(bool)]))
assert len(sensitivity_marked_dates) == 4
sensitivity_reference = float(holiday_coefficients[-1])
sensitivity_coefficients_before = holiday_coefficients.copy()
sensitivity_rows = []

for omitted_date in sensitivity_marked_dates:
    omitted_mask = sensitivity_training_dates == omitted_date
    retained_mask = ~omitted_mask
    assert omitted_mask.sum() == 24
    assert (holiday_train[omitted_mask] == 1).all()
    assert (maturity[train_mask][retained_mask] <= first_issue).all()
    assert set(sensitivity_training_dates[retained_mask]).isdisjoint({omitted_date})
    assert retained_mask[holiday_train == 0].all()
    remaining_marked_dates = set(
        sensitivity_training_dates[retained_mask & holiday_train.astype(bool)]
    )
    assert len(remaining_marked_dates) == 3
    sensitivity_fit, _, sensitivity_rank, _ = np.linalg.lstsq(
        X_train_holiday[retained_mask], y_train[retained_mask], rcond=None,
    )
    assert sensitivity_rank == X_train_holiday.shape[1]
    assert np.isfinite(sensitivity_fit).all()
    sensitivity_coefficient = float(sensitivity_fit[-1])
    sensitivity_rows.append({
        "Omitted Swiss local date": str(omitted_date),
        "Training hours": int(retained_mask.sum()),
        "Marked hours retained": int(holiday_train[retained_mask].sum()),
        "Holiday coefficient (MWh)": sensitivity_coefficient,
        "Change from full fit (MWh)": sensitivity_coefficient - sensitivity_reference,
        "Absolute change / full magnitude (%)": (
            100 * abs(sensitivity_coefficient - sensitivity_reference)
            / abs(sensitivity_reference)
        ),
    })

holiday_sensitivity = pd.DataFrame(sensitivity_rows).set_index("Omitted Swiss local date")
np.testing.assert_array_equal(holiday_coefficients, sensitivity_coefficients_before)
print(f"Full-training holiday coefficient: {sensitivity_reference:+.2f} MWh "
      f"({len(training):,} hours, {int(holiday_train.sum())} marked hours)")
display(holiday_sensitivity.round(2))
sensitivity_min = holiday_sensitivity["Holiday coefficient (MWh)"].min()
sensitivity_max = holiday_sensitivity["Holiday coefficient (MWh)"].max()
sensitivity_largest_change = holiday_sensitivity["Absolute change / full magnitude (%)"].max()
print(f"Omission-fit coefficient range: {sensitivity_min:+.2f} to {sensitivity_max:+.2f} MWh")
print(f"Largest absolute change relative to full-fit magnitude: {sensitivity_largest_change:.2f}%")
print("All omission-fit coefficients negative:",
      bool(holiday_sensitivity["Holiday coefficient (MWh)"].lt(0).all()))
```
